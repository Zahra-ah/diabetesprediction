# Create Model

```python
x=df.drop("Outcome",axis=1)
y=df.Outcome
```

# Splite the Data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,shuffle=True,random_state=42,test_size=0.2)
```

```python
x_test.loc[355]
```

```python
y_test
```

```python
model1=LogisticRegression()
model2=SVC()
model3=RandomForestClassifier(class_weight="balanced",n_estimators=200,n_jobs=-1)
model4=GradientBoostingClassifier()
model5=AdaBoostClassifier(n_estimators=100)
models=[model1,model2,model3,model4,model5]
```

```python
def cal(model):
    model.fit(x_train,y_train)
    preds=model.predict(x_test)
    accuracy=accuracy_score(preds,y_test)
    f1=f1_score(preds,y_test)
    recall=recall_score(preds,y_test)
    precision=precision_score(preds,y_test)
    print(model)
    print(f"recall is {recall} f1 is {f1} accuracy is {accuracy} precision is {precision} ")
    #print(classification_report(preds))
```

```python
for model in models:
    cal(model)
```
